# Classification Part A — Decision Tree Classifier

## Problem Statement

Predict the customer review score using order, product, geographic,
and engineered features from the e-commerce dataset.

### Algorithm
Decision Tree Classifier

### Evaluation Metrics
- Accuracy
- Precision
- Recall
- Weighted F1-Score
- Confusion Matrix
- ROC-AUC (One-vs-Rest)

```python
import numpy as np
import pandas as pd

from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = np.load("classification_data.npz")

X_train = data["X_train"]
X_test = data["X_test"]
y_train = data["y_train"]
y_test = data["y_test"]

feature_names = pd.read_csv(
    "classification_feature_names.csv",
    header=None
)[0].to_numpy()

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
print("Number of features:", len(feature_names))
```

## Decision Tree Classifier

A Decision Tree Classifier makes predictions by recursively splitting the
feature space based on feature values. It can model nonlinear relationships
and interactions between features.

The model is trained on the same prepared training data used by the other
Classification Part A algorithms.

```python
dt_model = DecisionTreeClassifier(
    random_state=42
)

print(dt_model)
```

### Model Training

The Decision Tree model is trained using only the prepared training
dataset. The test dataset remains unseen during training and is reserved for
final evaluation.

```python
dt_model.fit(X_train, y_train)

print("Decision Tree model trained successfully.")
```

### Predictions

The trained Gaussian Naive Bayes model is used to generate predictions for both
the training and unseen test datasets. The test predictions will be used for
the final model evaluation.

```python
y_train_pred = dt_model.predict(X_train)
y_test_pred = dt_model.predict(X_test)

print("Training predictions generated:", len(y_train_pred))
print("Test predictions generated:", len(y_test_pred))
```

## Model Evaluation

The Decision Tree Classifier is evaluated using accuracy, weighted precision,
weighted recall, weighted F1-score, and multiclass ROC-AUC using the
One-vs-Rest (OvR) approach.

```python
accuracy = accuracy_score(y_test, y_test_pred)

precision = precision_score(
    y_test,
    y_test_pred,
    average="weighted"
)

recall = recall_score(
    y_test,
    y_test_pred,
    average="weighted"
)

f1 = f1_score(
    y_test,
    y_test_pred,
    average="weighted"
)

roc_auc = roc_auc_score(
    y_test,
    dt_model.predict_proba(X_test),
    multi_class="ovr",
    average="weighted"
)

print("Accuracy       :", accuracy)
print("Precision      :", precision)
print("Recall         :", recall)
print("Weighted F1    :", f1)
print("ROC-AUC (OvR)  :", roc_auc)
```

## Confusion Matrix and Classification Report

The confusion matrix shows how the predicted review-score classes compare with
the actual classes. The classification report provides class-wise precision,
recall, and F1-score, helping identify which review-score classes are difficult
for the Decision Tree Classifier to predict.

```python
cm = confusion_matrix(y_test, y_test_pred)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Score 1", "Score 2", "Score 3", "Score 4", "Score 5"],
    yticklabels=["Score 1", "Score 2", "Score 3", "Score 4", "Score 5"]
)

plt.title("Confusion Matrix — Decision Tree Classifier")
plt.xlabel("Predicted Review Score")
plt.ylabel("Actual Review Score")
plt.tight_layout()
plt.show()

print("\nClassification Report:\n")

print(
    classification_report(
        y_test,
        y_test_pred,
        target_names=[
            "Score 1",
            "Score 2",
            "Score 3",
            "Score 4",
            "Score 5"
        ],
        zero_division=0
    )
)
```

### Observation

The Decision Tree Classifier shows similar precision, recall, and F1-scores
across the five review-score classes. Score 5 has the highest class-wise
F1-score (0.35), while Score 2 has the lowest (0.16). The model achieves an
accuracy of 0.26 and a weighted F1-score of 0.26, indicating relatively
uniform performance across the classes but limited overall predictive
performance.

```python
dt_results = {
    "Model": "Decision Tree Classifier",
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "Weighted F1": f1,
    "ROC-AUC": roc_auc
}

dt_results
```
