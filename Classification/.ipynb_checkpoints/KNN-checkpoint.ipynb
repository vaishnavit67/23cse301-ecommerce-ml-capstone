# Classification Part A — K-Nearest Neighbors (KNN)

## Problem Statement

Predict the customer review score using order, product, geographic,
and engineered features from the e-commerce dataset.

### Algorithm
K-Nearest Neighbors (KNN)

### Evaluation Metrics
- Accuracy
- Precision
- Recall
- Weighted F1-Score
- Confusion Matrix
- ROC-AUC (One-vs-Rest)

```python
import numpy as np
import pandas as pd

from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = np.load("classification_data.npz")

X_train = data["X_train"]
X_test = data["X_test"]
y_train = data["y_train"]
y_test = data["y_test"]

feature_names = pd.read_csv(
    "classification_feature_names.csv",
    header=None
)[0].to_numpy()

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
print("Number of features:", len(feature_names))
```

## K-Nearest Neighbors (KNN) Model

K-Nearest Neighbors (KNN) is a distance-based classification algorithm.
For each test instance, it identifies the nearest training instances and
assigns the class based on their labels.

Since the features have already been standardized during preprocessing,
distance calculations are performed on a comparable feature scale.

```python
knn_model = KNeighborsClassifier(
    n_neighbors=5
)

print(knn_model)
```

### Model Training

The KNN model is trained using the prepared training dataset. The test dataset
remains unseen during training and is used only for final evaluation.

```python
knn_model.fit(X_train, y_train)

print("KNN model trained successfully.")
```

### Predictions

The trained KNN model is used to generate predictions for both the training and
unseen test datasets. The test predictions will be used for the final model
evaluation.

```python
y_train_pred = knn_model.predict(X_train)
y_test_pred = knn_model.predict(X_test)

print("Training predictions generated:", len(y_train_pred))
print("Test predictions generated:", len(y_test_pred))
```

## Model Evaluation

The KNN model is evaluated on the unseen test set using accuracy, weighted
precision, weighted recall, and weighted F1-score. Weighted averaging is used
because the review-score classes are not perfectly balanced.

```python
accuracy = accuracy_score(y_test, y_test_pred)

precision = precision_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

print(f"Accuracy         : {accuracy:.4f}")
print(f"Weighted Precision: {precision:.4f}")
print(f"Weighted Recall   : {recall:.4f}")
print(f"Weighted F1-Score : {f1:.4f}")
```

```python
y_test_proba = knn_model.predict_proba(X_test)

roc_auc = roc_auc_score(
    y_test,
    y_test_proba,
    multi_class="ovr",
    average="weighted"
)

print(f"Weighted ROC-AUC (OvR): {roc_auc:.4f}")
```

## Confusion Matrix and Classification Report

The confusion matrix shows how the predicted review-score classes compare with
the actual classes. The classification report provides class-wise precision,
recall, and F1-score, helping identify which review-score classes are difficult
for the KNN model to predict.

```python
cm = confusion_matrix(y_test, y_test_pred)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[1, 2, 3, 4, 5],
    yticklabels=[1, 2, 3, 4, 5]
)

plt.title("KNN - Confusion Matrix")
plt.xlabel("Predicted Review Score")
plt.ylabel("Actual Review Score")
plt.show()

print("Classification Report:")
print(
    classification_report(
        y_test,
        y_test_pred,
        target_names=["Score 1", "Score 2", "Score 3", "Score 4", "Score 5"],
        zero_division=0
    )
)
```

### Observation

The KNN model shows relatively low and balanced performance across all five
review-score classes. Score 3 has the highest class-wise F1-score (0.30),
while Score 1 and Score 2 have the lowest F1-scores (0.19). The weighted
F1-score of 0.26 is close to the accuracy of 0.26, indicating that no single
class dominates the overall performance. Overall, KNN performs weaker than
the Logistic Regression baseline on this dataset.

```python
knn_results = {
    "Model": "KNN",
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "Weighted F1": f1,
    "ROC-AUC": roc_auc
}

knn_results
```
