# Classification Part A — Logistic Regression

## Problem Statement

Predict the customer review score using order, product, geographic, 
and engineered features from the e-commerce dataset.

### Algorithm
Logistic Regression

### Evaluation Metrics
- Accuracy
- Precision
- Recall
- Weighted F1-Score
- Confusion Matrix
- ROC-AUC (One-vs-Rest)

```python
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score
)

import matplotlib.pyplot as plt
import seaborn as sns
```

## Load Prepared Classification Data

The cleaned, feature-engineered, stratified train-test split generated during
the EDA and preprocessing stage is loaded here. All classification algorithms
use this same prepared dataset to ensure a consistent and fair comparison.

```python
data = np.load("classification_data.npz")

X_train = data["X_train"]
X_test = data["X_test"]
y_train = data["y_train"]
y_test = data["y_test"]

feature_names = pd.read_csv(
    "classification_feature_names.csv",
    header=None
)[0].to_numpy()

print("X_train shape:", X_train.shape)
print("X_test shape :", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape :", y_test.shape)
print("Number of features:", len(feature_names))
```

## Logistic Regression Model

Logistic Regression is used as the first baseline classification algorithm.
Since the target variable contains five classes (review scores 1–5), the model
performs multiclass classification.

The model is trained on the preprocessed training data and evaluated later on
the unseen test data.

```python
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

print(logistic_model)
```

### Model Training

The Logistic Regression model is trained using only the prepared training
dataset. The test dataset remains completely unseen during training and will
be used only for final evaluation.

```python
logistic_model.fit(X_train, y_train)

print("Logistic Regression model trained successfully.")
```

### Predictions

The trained Logistic Regression model is used to predict the review-score
class for both the training and unseen test datasets. The test predictions
will be used for the final evaluation.

```python
y_train_pred = logistic_model.predict(X_train)
y_test_pred = logistic_model.predict(X_test)

print("Training predictions generated:", len(y_train_pred))
print("Test predictions generated:", len(y_test_pred))
```

## Model Evaluation

The Logistic Regression model is evaluated on the unseen test set using
accuracy, weighted precision, weighted recall, and weighted F1-score.
Weighted averaging is used because the review-score classes are not perfectly
balanced.

```python
accuracy = accuracy_score(y_test, y_test_pred)

precision = precision_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_test_pred,
    average="weighted",
    zero_division=0
)

print(f"Accuracy         : {accuracy:.4f}")
print(f"Weighted Precision: {precision:.4f}")
print(f"Weighted Recall   : {recall:.4f}")
print(f"Weighted F1-Score : {f1:.4f}")
```

```python
y_test_proba = logistic_model.predict_proba(X_test)

roc_auc = roc_auc_score(
    y_test,
    y_test_proba,
    multi_class="ovr",
    average="weighted"
)

print(f"Weighted ROC-AUC (OvR): {roc_auc:.4f}")
```

## Confusion Matrix and Classification Report

The confusion matrix shows how the predicted review-score classes compare with
the actual classes. The classification report provides class-wise precision,
recall, and F1-score, which helps identify classes that are difficult for the
Logistic Regression model to predict.

```python
# Confusion matrix
cm = confusion_matrix(y_test, y_test_pred)

plt.figure(figsize=(7, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=[1, 2, 3, 4, 5],
    yticklabels=[1, 2, 3, 4, 5]
)

plt.title("Logistic Regression - Confusion Matrix")
plt.xlabel("Predicted Review Score")
plt.ylabel("Actual Review Score")
plt.show()


# Class-wise classification report
print("Classification Report:")
print(
    classification_report(
        y_test,
        y_test_pred,
        target_names=["Score 1", "Score 2", "Score 3", "Score 4", "Score 5"],
        zero_division=0
    )
)
```

## Logistic Regression Results

The Logistic Regression model provides the baseline performance for the
classification task. Its results will be compared with the other Part-A
classification algorithms using the same prepared train-test split.

```python
logistic_results = {
    "Model": "Logistic Regression",
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "Weighted F1": f1,
    "ROC-AUC": roc_auc
}

logistic_results
```
